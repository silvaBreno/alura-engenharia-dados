````python
import pandas as pd
import re
from calendar import monthrange
from datetime import datetime

# 1. Dicionário de Mapeamento de Meses (em minúsculas)
meses_map = {
    'janeiro': '01', 'fevereiro': '02', 'março': '03', 'abril': '04',
    'maio': '05', 'junho': '06', 'julho': '07', 'agosto': '08',
    'setembro': '09', 'outubro': '10', 'novembro': '11', 'dezembro': '12'
}

# ----------------------------------------------------------------------
# FUNÇÕES AUXILIARES DE PRÉ-PROCESSAMENTO
# ----------------------------------------------------------------------

def _preparar_periodo(periodo):
    """Limpa e normaliza o período, retornando a string e o valor numérico."""
    periodo_limpo = str(periodo).lower().strip().replace('º', '')
    
    # 2.1. Tratamento de valores inválidos que devem ser ignorados
    if periodo_limpo in ['diversos', '---------------', 'none', 'n/a']:
        return None, None
    
    # Prepara o período numérico (substitui nomes de meses por números)
    periodo_numerico = periodo_limpo
    for pt, num in meses_map.items():
        # Usa word boundary (\b) para evitar substituições parciais
        periodo_numerico = re.sub(r'\b' + pt + r'\b', num, periodo_numerico)
        
    return periodo_limpo, periodo_numerico

# ----------------------------------------------------------------------
# FUNÇÕES AUXILIARES DE TRATAMENTO DE PADRÕES
# ----------------------------------------------------------------------

def _tratar_ano_calendario(periodo_original):
    """2.2. Tratamento de Ano-calendário (Ex: 'Ano-calendário de 2024')"""
    match_ano = re.search(r'202\d', periodo_original)
    if match_ano and ('ano-calendário' in periodo_original or 'exercício' in periodo_original):
        ano = int(match_ano.group(0))
        return pd.to_datetime(f'01/01/{ano}', format='%d/%m/%Y'), pd.to_datetime(f'31/12/{ano}', format='%d/%m/%Y')
    return pd.NaT, pd.NaT

def _tratar_mes_ano_simples(periodo_numerico):
    """2.3. Mês/Ano Simples (Ex: 05/2025 ou 02/20255 - com a correção)"""
    # Permite de 4 a 5 dígitos no ano para lidar com erros de digitação (ex: 20255)
    match_mes_ano = re.search(r'(\d{2})/?(\d{4,5})$', periodo_numerico) 
    if match_mes_ano:
        mes = int(match_mes_ano.group(1))
        
        # Corrige o ano, pegando apenas os 4 primeiros dígitos
        ano_str = match_mes_ano.group(2)
        ano = int(ano_str[:4]) 
        
        data_inicio = pd.to_datetime(f'01/{mes}/{ano}', format='%d/%m/%Y', errors='coerce')
        
        try:
            # Obtém o último dia do mês com precisão
            dias_no_mes = monthrange(ano, mes)[1] 
            data_fim = pd.to_datetime(f'{dias_no_mes}/{mes}/{ano}', format='%d/%m/%Y', errors='coerce')
            return data_inicio, data_fim
        except ValueError:
            return pd.NaT, pd.NaT # Inválido se o mês ou ano for maluco
    return pd.NaT, pd.NaT

def _tratar_intervalo_mes_a_mes(periodo_numerico):
    """2.4. Intervalo de Mês a Mês/Ano (Ex: 10 a 12/2024)"""
    match_intervalo_meses_simples = re.search(r'(\d{2})\s?a\s?(\d{2})/(\d{4})', periodo_numerico)
    if match_intervalo_meses_simples:
        mes_inicio_num, mes_fim_num, ano = match_intervalo_meses_simples.groups()
        
        ano_int = int(ano)
        mes_fim_int = int(mes_fim_num)
        
        data_inicio = pd.to_datetime(f'01/{mes_inicio_num}/{ano}', format='%d/%m/%Y', errors='coerce')
        
        try:
            dias_no_mes_fim = monthrange(ano_int, mes_fim_int)[1]
            data_fim = pd.to_datetime(f'{dias_no_mes_fim}/{mes_fim_num}/{ano}', format='%d/%m/%Y', errors='coerce')
            return data_inicio, data_fim
        except ValueError:
            return pd.NaT, pd.NaT
    return pd.NaT, pd.NaT

def _tratar_intervalo_completo(periodo_numerico):
    """2.5. Intervalo de Data Completa (Ex: 1/10 a 31/12/2024)"""
    # Padrão: dia/mês a dia/mês/ano (usando os números dos meses)
    match_intervalo_completo = re.search(r'(\d{1,2})\s?(\d{2})\s?a\s?(\d{1,2})\s?(\d{2})/(\d{4})', periodo_numerico)
    if match_intervalo_completo:
        dia_inicio, mes_inicio, dia_fim, mes_fim, ano = match_intervalo_completo.groups()
        
        data_inicio = pd.to_datetime(f'{dia_inicio}/{mes_inicio}/{ano}', format='%d/%m/%Y', errors='coerce')
        data_fim = pd.to_datetime(f'{dia_fim}/{mes_fim}/{ano}', format='%d/%m/%Y', errors='coerce')
        return data_inicio, data_fim
    return pd.NaT, pd.NaT

def _tratar_intervalo_dias_mesmo_mes(periodo_numerico):
    """2.6. Intervalo de Dias em um Único Mês (Ex: 1 a 10/03/2025)"""
    match_intervalo_dias = re.search(r'(\d{1,2})\s?a\s?(\d{1,2})/(\d{2})/(\d{4})', periodo_numerico)
    if match_intervalo_dias:
        dia_inicio, dia_fim, mes, ano = match_intervalo_dias.groups()
        data_inicio = pd.to_datetime(f'{dia_inicio}/{mes}/{ano}', format='%d/%m/%Y', errors='coerce')
        data_fim = pd.to_datetime(f'{dia_fim}/{mes}/{ano}', format='%d/%m/%Y', errors='coerce')
        return data_inicio, data_fim
    return pd.NaT, pd.NaT

def _tratar_intervalo_dias_mes_extenso(periodo_limpo):
    """2.7. Intervalo de Dias com Mês por Extenso (Ex: 21 a 31/dezembro/2024)"""    
    match = re.search(r'(\d{1,2})\s?a\s?(\d{1,2})/([a-zç]+)/(\d{4})', periodo_limpo)
    if match:
        dia_inicio, dia_fim, mes_extenso, ano = match.groups()
        mes = meses_map.get(mes_extenso)
        if mes:
            data_inicio = pd.to_datetime(f'{dia_inicio}/{mes}/{ano}', format='%d/%m/%Y', errors='coerce')
            data_fim = pd.to_datetime(f'{dia_fim}/{mes}/{ano}', format='%d/%m/%Y', errors='coerce')
            return data_inicio, data_fim
    return pd.NaT, pd.NaT


# ----------------------------------------------------------------------
# FUNÇÃO MESTRE ORQUESTRADORA
# ----------------------------------------------------------------------

def padronizar_periodo(periodo):
    """
    Função principal que tenta padronizar o periodo_fato_gerador em 
    data_inicio e data_fim, usando funções auxiliares.
    """
    if pd.isna(periodo):
        return pd.NaT, pd.NaT
        
    # Passo 1: Limpeza e Checagem de Invalidade
    periodo_limpo, periodo_numerico = _preparar_periodo(periodo)
    
    if periodo_limpo is None: # Se foi tratado como 'Diversos', '---------------', etc.
        return pd.NaT, pd.NaT

    # Passo 2: Orquestração. Tenta em ordem de complexidade
    
    # Tenta o Intervalo de Dias com Mês por Extenso (Ex: 21 a 31/dezembro/2024)    
    data_inicio, data_fim = _tratar_intervalo_dias_mes_extenso(periodo_limpo)
    if data_inicio is not pd.NaT:
        return data_inicio, data_fim

    # Tenta o Intervalo de Dias em um Único Mês (Ex: 1 a 10/03/2025)
    data_inicio, data_fim = _tratar_intervalo_dias_mesmo_mes(periodo_numerico)
    if data_inicio is not pd.NaT:
        return data_inicio, data_fim
    
    # Tenta o Intervalo de Mês a Mês (Ex: Outubro a Dezembro/2024)
    data_inicio, data_fim = _tratar_intervalo_mes_a_mes(periodo_numerico)
    if data_inicio is not pd.NaT:
        return data_inicio, data_fim


    # Tenta o Intervalo de Data Completa (Ex: 1 Outubro a 31 Dezembro/2024)
    data_inicio, data_fim = _tratar_intervalo_completo(periodo_numerico)
    if data_inicio is not pd.NaT:
        return data_inicio, data_fim

    # Tenta o Mês/Ano Simples (Ex: Maio/2025) - Mais comum e inclui a correção de dígitos
    data_inicio, data_fim = _tratar_mes_ano_simples(periodo_numerico)
    if data_inicio is not pd.NaT:
        return data_inicio, data_fim
        
    # Tenta o Ano-Calendário (Ex: Ano-calendário de 2024) - Mantido por último pela sua especificidade textual
    data_inicio, data_fim = _tratar_ano_calendario(periodo_limpo)
    if data_inicio is not pd.NaT:
        return data_inicio, data_fim

    # Passo 3: Não Padronizado
    return pd.NaT, pd.NaT

# ----------------------------------------------------------------------
# EXEMPLO DE APLICAÇÃO
# ----------------------------------------------------------------------

df[['data_inicio', 'data_fim']] = df['periodo_fato_gerador'].apply(
     lambda x: pd.Series(padronizar_periodo(x))
)
````
