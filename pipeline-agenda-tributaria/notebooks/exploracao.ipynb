### Importando as bibliotecas

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import json
import time
from io import StringIO
```

```python
# URL base da agenda tribunária
url_base = "https://www.gov.br/receitafederal/pt-br/assuntos/agenda-tributaria"

# Ano selecionado pelo usuário
ano_selecionado = 2025

# URL com o ano selecionado
url_com_ano = f"{url_base}/{ano_selecionado}"
url_com_ano
```

```python
# Realizando uma requisição GET na URL base da agenda tributária
response = requests.get(url_com_ano)
response.status_code
```

```python
# analisando o conteudo 
response.content
```

```python
# analisando o conteudo deixando a mensagem do terminal mais legível
soup = BeautifulSoup(response.content, 'html.parser')
print(soup.prettify())
type(soup)
```

```python
# Extraindo o link dos meses
links_meses = []
for a in soup.find_all('a', href=True):
    if f"/{ano_selecionado}/" in a['href'] and any(mes in a['href'] for mes in ['janeiro', 'fevereiro', 'março', 'abril', 'maio', 'junho', 'julho', 'agosto', 'setembro', 'outubro', 'novembro', 'dezembro']):
        links_meses.append(a['href'])
links_meses
```

```python
links_meses = [
        a['href'] for a in soup.find_all('a', href=True)
        if a['href'].startswith(f"{url_com_ano}/") and any(m in a['href'] for m in ['janeiro', 'fevereiro', 'marco', 'abril', 'maio', 'junho', 'julho', 'agosto', 'setembro', 'outubro', 'novembro', 'dezembro'])
    ]
links_meses
```

```python
# Preciso agora pegar os links das datas com os eventos em cada mes
print(links_meses[0])
## Vou pegar manualmente o primeiro mes
response = requests.get(links_meses[0])
print(f"URL sendo testada: {response.url}\nStatus code: {response.status_code}\n")
response.status_code
soup = BeautifulSoup(response.content, 'html.parser')
print(soup.prettify())
```

```python
print(response.url)
# janeiro
links_datas = []
for a in soup.select('a'):
    if a['href'].startswith(response.url) and '/dia-' in a['href']:
        links_datas.append(a['href'])

print(len(links_datas))
print(links_datas)
```

```python
# Dicionário para armazenar os dados
dados_agenda = {}
nome_mes = 'janeiro'
dados_agenda[nome_mes] = {}
print(dados_agenda)
```

```python
for link_data in links_datas:
    nome_data = link_data.split('/')[-1]  # Extrai o nome da data do link
    print(nome_data)
    url_data = link_data
    print(url_data)
    response = requests.get(url_data)
    soup = BeautifulSoup(response.content, 'html.parser')
    tabelas_html = soup.find_all('table')
    tabelas = []
    for tabela in tabelas_html:
        df = pd.read_html(StringIO(str(tabela)))[0]  # Corrigido
        tabelas.append(df.to_dict(orient='records'))
    dados_agenda[nome_mes][nome_data] = tabelas
    print(dados_agenda)
```

```python
# salvando os dados em formato JSON
with open(f'../data/raw/exploracao_dados_agenda_tributaria_janeiro_{ano_selecionado}.json', 'w', encoding='utf-8') as f:
    json.dump(dados_agenda, f, ensure_ascii=False, indent=4)

print('Dados salvos em exploracao_dados_agenda_tributaria.json')
```
